```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from termcolor import cprint
import results_analysis_utils as rutils
from IPython.display import display


%load_ext autoreload
%autoreload 2

# ===================================================================
# 1. CONFIGURATION
# ===================================================================
day_code = "2024_11_12" 
dataset_name = "DATASET_1" # O DATASET_2

ASPECT_RATIO_THR = 3.0
ANGLE_THR = 30.0

# Ruta al CSV crudo de este día concreto
csv_path = Path(f"/home/slimbook/fish_sizing/ARTICLE/{dataset_name}/{day_code}/{day_code}_ALL_raw_aggregated.csv")

cprint(f"👉 WORKING ON: {day_code}", "white", "on_blue", attrs=["bold"])
cprint(f"Reading csv from {csv_path}", "yellow")

# ===================================================================
# 2. FILTERING & AGGREGATION (Usando la misma lógica global)
# ===================================================================
# Le pasamos el nombre del día como 'dataset_name' para que la función imprima bonito

df_final, df_raw = rutils.load_and_filter_dataset(csv_path, 
                                                  dataset_name=day_code,
                                                  recalculate_failure_reason=True, 
                                                  aspect_ratio_thr=ASPECT_RATIO_THR, 
                                                  angle_thr=ANGLE_THR, 
                                                  save=False)

if df_raw.empty:
    raise ValueError("No data found for this day!")

# ===================================================================
# 3. DAILY SUMMARY TABLE (Single vs Multiple)
# ===================================================================
print("\n" + "="*85)
print(f"DAILY SUMMARY: {day_code}")
print("="*85)
print(f"{'SCENARIO':<15} | {'Total Tracks':<12} | {'Measured':<10} | {'Success (%)':<12} | {'MAE (cm)':<10} | {'STD (cm)':<10} | {'MAPE (%)':<10}")
print("-" * 85)

scenarios = ['All']
if 'scenario' in df_raw.columns:
    scenarios.extend(df_raw['scenario'].dropna().unique().tolist())

for sc in scenarios:
    # 1. Contar totales desde el RAW
    if sc == 'All':
        total_tracks = df_raw['track_uid'].nunique()
        sub_final = df_final
    else:
        total_tracks = df_raw[df_raw['scenario'] == sc]['track_uid'].nunique()
        sub_final = df_final[df_final['scenario'] == sc] if not df_final.empty else pd.DataFrame()
            
    # 2. Contar medidos desde el FINAL
    measured_tracks = len(sub_final)
    
    # 3. Calcular métricas
    if measured_tracks > 0:
        mae = sub_final['abs_error_cm'].mean()
        std = sub_final['abs_error_cm'].std()
        mape = sub_final['rel_error_perc'].mean()
    else:
        mae = std = mape = 0.0
        
    success_perc = (measured_tracks / total_tracks * 100) if total_tracks > 0 else 0
    
    label = f"GLOBAL DAY" if sc == 'All' else sc
    print(f"{label:<15} | {total_tracks:<12} | {measured_tracks:<10} | {success_perc:<12.1f} | {mae:<10.2f} | {std:<10.2f} | {mape:<10.2f}")

print("="*85 + "\n")


for sc in scenarios:
    cprint(f"\n\n{'='*70}", "white", "on_magenta", attrs=["bold"])
    cprint(f"📊 PLOTS FOR SCENARIO: {sc.upper()}", "white", "on_magenta", attrs=["bold"])
    cprint(f"{'='*70}", "white", "on_magenta", attrs=["bold"])
    
    # 1. Filtrar el dataframe dependiendo del escenario
    if sc == 'All':
        df_sc_raw = df_raw.copy()
        context_label = f"{day_code} - GLOBAL"
    else:
        df_sc_raw = df_raw[df_raw['scenario'] == sc].copy()
        context_label = f"{day_code} - {sc.upper()}"
        
    if df_sc_raw.empty:
        print(f"⚠️ No data available for {sc.upper()}, skipping plots.")
        continue
    
    # 2. Generar gráficos
    stacked_frames = rutils.plot_frame_failures(
        df_sc_raw, 
        aspect_ratio_thr=ASPECT_RATIO_THR, 
        angle_thr=ANGLE_THR,
        figsize_video=(12,6), 
        figsize_global=(5,5), 
        show_global=True,
        context_label=context_label
    )

    track_failures = rutils.plot_tracks_failure_distribution(
        df_sc_raw, 
        aspect_ratio_thr=ASPECT_RATIO_THR, 
        angle_thr=ANGLE_THR,
        show_global=True, 
        figsize_video=(12,6), 
        figsize_global=(5,5),
        context_label=context_label
    )

    rutils.plot_aspect_ratio_vs_length(
        df_sc_raw, 
        ASPECT_RATIO_THR=ASPECT_RATIO_THR, 
        figsize=(10, 8),
        context_label=context_label
    ) 
    
    # 1. Gráfica de Cascada (Ablation)
    print("\n📈 Generating Pipeline Ablation Plots...")
    stages_dict, agg_stages_dict = rutils.plot_ablation_pipeline(
        df_sc_raw, 
        ASPECT_RATIO_THR=ASPECT_RATIO_THR, 
        ANGLE_THR=ANGLE_THR,
        context_label=context_label  
    )
    
    # 2. Interacción de umbrales
    print("\n📈 Generating Thresholds Interaction Plot...")
    rutils.plot_thresholds_interaction(
        df_sc_raw, 
        optimal_ar=ASPECT_RATIO_THR, 
        context=context_label
    )
    
    # 3. Relación de Error vs Geometría (Raw y Aislado)
    print("\n📈 Generating Error vs Geometry Analysis Plots...")
    rutils.plot_error_vs_geometry(
        df_sc_raw, 
        ar_thr=ASPECT_RATIO_THR, 
        angle_thr=ANGLE_THR, 
        context_label=context_label
    )
    
    print("\n📈 Generating Track Error vs Length Plot...")
    rutils.plot_error_vs_track_length(
        df_raw=df_sc_raw, 
        df_final=sub_final,  # <--- ¡AQUÍ LE PASAMOS TU CÁLCULO INTELIGENTE!
        context_label=context_label,
        min_frames_thr=5
    )
    
    print("\n📈 Generating Track Length Histogram...")
    rutils.plot_track_length_histogram(
        df_sc_raw, 
        min_frames_thr=5, # O el número que uses en el smart_aggregator
        context_label=context_label
    )
```
